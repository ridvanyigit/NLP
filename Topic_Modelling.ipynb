```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import ktrain
import warnings
warnings.filterwarnings('ignore')
```

```python
ktrain.__version__
```

```python
from sklearn.datasets import fetch_20newsgroups
```

```python
remove = ('headers', 'footers', 'quotes')
data = fetch_20newsgroups(subset='train', shuffle=True, random_state=42, remove=remove)
```

```python
texts = data.data
texts[:15]  # Display the first 15 texts
```

```python
targets = data.target
targets
```

```python
len(texts), len(targets)
```

```python
data.target_names
```

```python
categories = [data.target_names[target] for target in targets]
```

```python
%%time
tm = ktrain.text.get_topic_model(texts, n_features=10000)
#tm = topic model
```

```python
tm.print_topics()
```

```python
%%time
tm.build(texts, threshold=0.25)
```

```python
texts = tm.filter(texts)
```

```python
categories = tm.filter(categories)
```

```python
tm.print_topics(show_counts=True)
```

```python
tm.visualize_documents(doc_topics= tm.get_doctopics())
```

```python
doc_topics= tm.get_doctopics(topic_ids=[50]) # topic:50 | count:910 | drive card disk hard mac bit video scsi use drives
tm.visualize_documents(doc_topics= doc_topics)
```

```python
news = '''
Drives encompass various storage solutions like hard disks, SSDs (solid state drives), and even older technologies like SCSI.
On Macs, drives store everything from operating system bits to video files.
Choosing the right type, capacity, and speed is crucial for optimal performance, whether it's a disk, card, or other storage use.
'''
```

```python
tm.predict([news])
```

```python
topics = tm.get_topics()

for topic_index, topic in enumerate(topics):
    if topic_index == 50:
        print(f"topic:{topic_index} {topic}")
        break
```

```python
pred_topic = tm.topics[np.argmax(tm.predict([news]))]
pred_topic
```
